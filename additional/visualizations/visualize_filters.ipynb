```python
from flower_model.external._gabor_like_filters import _gabor_like_filters
import matplotlib.pyplot as plt
import numpy as np
```

```python
N = 27
sigma = 15.
steps = 11
percent = 0.7
device = 'cpu'

filters, _ = _gabor_like_filters(N, sigma, steps, percent)
print(filters.shape)
```

```python
filter_amount = filters.shape[0]

t_steps = 27
r_steps = int((filter_amount - 1) / t_steps)
```

```python
fig, axes = plt.subplots(t_steps, r_steps, figsize=(r_steps * 1.5, t_steps * 1.5))
for t in range(t_steps):
    for r in range(r_steps):
        idx = t * r_steps + r + 1
        axes[t, r].imshow(np.abs(filters)[idx], cmap="viridis")
        axes[t, r].axis("off")
plt.tight_layout()
plt.show()
```
