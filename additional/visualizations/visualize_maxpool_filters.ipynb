```python
from flower_model._filterbank import _create_filterbank
import matplotlib.pyplot as plt
```

```python
kernel_size = 127
sigma = 150.
k = 100
device = "cpu"

filters_normalized, frequency_domain_sum = _create_filterbank(
    N=kernel_size, sigma=sigma, k=k, device=device
)

filters_normalized = filters_normalized.view(-1, kernel_size, kernel_size)

print(filters_normalized.shape)
```

```python
r_steps = 10
t_steps = int(k*4/r_steps)

fig, axes = plt.subplots(t_steps, r_steps, figsize=(r_steps * 1.5, t_steps * 1.5))
for t in range(t_steps):
    for r in range(r_steps):
        idx = t * r_steps + r
        axes[t, r].imshow(filters_normalized[idx], cmap="viridis")
        axes[t, r].axis("off")
plt.tight_layout()
plt.show()
```
