```python
import matplotlib.pyplot as plt
import math
import numpy as np
import torch
```

```python
features_new_test = torch.load("")
print(features_new_test.shape)
```

```python
image = features_new_test[2].view(1, 114, -1)

print(image[0][8])

plt.imshow(image[0][8].view(16,16))
plt.show()

f = image.squeeze(0).cpu()        # shape [N, 1024]
print(f.shape)
n = f.shape[0]
size = int(math.sqrt(f.shape[1]))
assert size * size == f.shape[1], "expected square (H*W)"

cols = int(math.ceil(math.sqrt(n)))
rows = int(math.ceil(n / cols))

fig, axes = plt.subplots(rows, cols, figsize=(cols*1.5, rows*1.5))
axes = np.atleast_1d(axes).flatten()

for i in range(rows * cols):
    ax = axes[i]
    if i < n:
        img = f[i].reshape(size, size).numpy()
        ax.imshow(img, cmap="gray")
    ax.axis("off")

plt.tight_layout()
plt.show()
```
